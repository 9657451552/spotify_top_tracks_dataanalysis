```python
import pandas as pd

# Load dataset
df = pd.read_csv("top2018.csv")

# Display first 5 rows
df.head()
```

```python
# Check dataset size
print("Dataset Shape:", df.shape)

# Check column names
print("\nColumn Names:")
print(df.columns)

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Statistical summary of numerical columns
df.describe()
```

```python
# Calculate correlation between numerical features
correlation = df.corr(numeric_only=True)

# Display correlation matrix
correlation
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))

plt.imshow(correlation, cmap="coolwarm", aspect="auto")

plt.colorbar(label="Correlation")

plt.xticks(range(len(correlation.columns)),
           correlation.columns, rotation=90)

plt.yticks(range(len(correlation.columns)),
           correlation.columns)

plt.title("Correlation Heatmap of Spotify Audio Features")

plt.tight_layout()
plt.show()
```

```python
# Count songs by each artist
artist_counts = df["artists"].value_counts()

# Display top 10 artists
artist_counts.head(10)
```

```python
import matplotlib.pyplot as plt

top_artists = artist_counts.head(10)

plt.figure(figsize=(10, 6))
top_artists.plot(kind="bar")

plt.title("Top 10 Artists by Number of Songs")
plt.xlabel("Artists")
plt.ylabel("Number of Songs")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Display top 10 songs
top_songs = df[["name", "artists", "energy", "danceability"]].head(10)

top_songs
```

```python
# Top 10 songs with highest energy
top_energy = df[["name", "artists", "energy"]].sort_values(
    by="energy", ascending=False
).head(10)

top_energy
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.barh(top_energy["name"], top_energy["energy"])

plt.title("Top 10 Songs by Energy")
plt.xlabel("Energy")
plt.ylabel("Song")

plt.gca().invert_yaxis()
plt.tight_layout()

plt.show()
```

```python
print("Average Danceability:", round(df["danceability"].mean(), 3))
print("Average Energy:", round(df["energy"].mean(), 3))
print("Average Loudness:", round(df["loudness"].mean(), 3))
print("Average Tempo:", round(df["tempo"].mean(), 2), "BPM")
print("Average Valence:", round(df["valence"].mean(), 3))
print("Average Duration:", round(df["duration_ms"].mean() / 60000, 2), "minutes")
```

This project analyzed the Spotify Top Tracks dataset using Python and explored different audio features such as danceability, energy, loudness, acousticness, valence, and tempo. The dataset was checked for missing values and statistical characteristics before performing exploratory data analysis and visualization.

The analysis identified highly danceable and high-energy songs and examined relationships between different audio features. A strong positive relationship was observed between energy and loudness, while acousticness showed a negative relationship with energy. Tempo and danceability showed a weak negative relationship.

Overall, the project demonstrates how data analysis and visualization can convert raw music data into meaningful insights that can support playlist creation, song analysis, and recommendation-related decisions.
